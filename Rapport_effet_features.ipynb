**Conclusion générale:** Les **vitesses du peg** seules donnent une **convergence lente et instable** (plateau + pics) sur 40–80 époques, mais avec **150 époques** le modèle finit par **séparer parfaitement** les classes (**100% test, loss 0.006**).
